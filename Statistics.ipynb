```python
import mysql.connector
from getpass import getpass

password = getpass("Enter MySQL password: ")

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password=password,
    database="student_learning_db"
)

print("Connected successfully!")
```

```python
import pandas as pd

query = "SELECT * FROM student_learning_data"

df = pd.read_sql(query, connection)

print("Data loaded successfully!")
print(df.shape)
```

```python
df.head()
```

```python
# Statistics Step 1: Mean, Median and Mode

mean_readiness = df['overall_readiness_score'].mean()
median_readiness = df['overall_readiness_score'].median()
mode_readiness = df['overall_readiness_score'].mode()[0]

print("Mean Readiness Score:", mean_readiness)
print("Median Readiness Score:", median_readiness)
print("Mode Readiness Score:", mode_readiness)
```

```python
# Statistics Step 2: Range, Variance and Standard Deviation

range_readiness = (
    df['overall_readiness_score'].max()
    - df['overall_readiness_score'].min()
)

variance_readiness = df['overall_readiness_score'].var()
std_readiness = df['overall_readiness_score'].std()

print("Minimum Readiness Score:", df['overall_readiness_score'].min())
print("Maximum Readiness Score:", df['overall_readiness_score'].max())
print("Range:", range_readiness)
print("Variance:", variance_readiness)
print("Standard Deviation:", std_readiness)
```

```python
# Statistics Step 3: Quartiles and IQR

Q1 = df['overall_readiness_score'].quantile(0.25)
Q2 = df['overall_readiness_score'].quantile(0.50)
Q3 = df['overall_readiness_score'].quantile(0.75)

IQR = Q3 - Q1

print("Q1 (25th percentile):", Q1)
print("Q2 (50th percentile / Median):", Q2)
print("Q3 (75th percentile):", Q3)
print("IQR:", IQR)
```

```python
# Statistics Step 4: Outlier Detection using IQR

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['overall_readiness_score'] < lower_bound) |
    (df['overall_readiness_score'] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

```python
outliers[['student_id', 'overall_readiness_score', 'risk_level', 'placement_ready']].head(20)
```

```python
# Statistics Step 5: Skewness

skewness = df['overall_readiness_score'].skew()

print("Skewness of Overall Readiness Score:", skewness)
```

```python
# Statistics Step 6: Kurtosis

kurtosis = df['overall_readiness_score'].kurt()

print("Kurtosis of Overall Readiness Score:", kurtosis)
```

```python
# Statistics Step 7: Standard Error

standard_error = df['overall_readiness_score'].sem()

print("Standard Error of Mean:", standard_error)
```

```python
# Statistics Step 8: 95% Confidence Interval

mean = df['overall_readiness_score'].mean()
se = df['overall_readiness_score'].sem()

lower_ci = mean - 1.96 * se
upper_ci = mean + 1.96 * se

print("Mean Readiness Score:", mean)
print("95% Confidence Interval:")
print("Lower Bound:", lower_ci)
print("Upper Bound:", upper_ci)
```

```python
# Statistics Step 9: One-Sample T-Test

from scipy.stats import ttest_1samp

test_result = ttest_1samp(
    df['overall_readiness_score'],
    popmean=60
)

print("T-statistic:", test_result.statistic)
print("P-value:", test_result.pvalue)
```

```python
# Statistics Step 10: Final Statistics Summary

readiness = df['overall_readiness_score']

print("===== FINAL STATISTICS SUMMARY =====")

print("Mean:", readiness.mean())
print("Median:", readiness.median())
print("Mode:", readiness.mode()[0])

print("Minimum:", readiness.min())
print("Maximum:", readiness.max())
print("Range:", readiness.max() - readiness.min())

print("Variance:", readiness.var())
print("Standard Deviation:", readiness.std())

print("Q1:", readiness.quantile(0.25))
print("Q2 (Median):", readiness.quantile(0.50))
print("Q3:", readiness.quantile(0.75))

print("IQR:", readiness.quantile(0.75) - readiness.quantile(0.25))

print("Skewness:", readiness.skew())
print("Kurtosis:", readiness.kurtosis())

print("Standard Error:", readiness.sem())

print("95% CI Lower:", readiness.mean() - 1.96 * readiness.sem())
print("95% CI Upper:", readiness.mean() + 1.96 * readiness.sem())
```
