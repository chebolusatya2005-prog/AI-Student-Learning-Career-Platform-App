```python
def readiness_category(score):
    if score < 40:
        return 'Below 40'
    elif score < 60:
        return '40-59'
    elif score < 80:
        return '60-79'
    else:
        return '80+'

df['readiness_category'] = df['overall_readiness_score'].apply(
    readiness_category
)

df['readiness_category'].value_counts()
```

```python
readiness_counts = df['readiness_category'].value_counts()

plt.figure(figsize=(8, 5))

readiness_counts.plot(kind='bar')

plt.title('Students by Overall Readiness Range')
plt.xlabel('Readiness Range')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
df['attendance_group'] = pd.cut(
    df['attendance_percentage'],
    bins=[0, 50, 70, 85, 100],
    labels=['Below 50%', '50-70%', '70-85%', '85%+']
)

attendance_analysis = df.groupby(
    'attendance_group',
    observed=True
)['overall_readiness_score'].agg(
    ['count', 'mean']
)

attendance_analysis
```

```python
attendance_analysis['mean'].plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Average Readiness Score by Attendance Group')
plt.xlabel('Attendance Group')
plt.ylabel('Average Readiness Score')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
